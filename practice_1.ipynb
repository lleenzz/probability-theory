### Пример 1.10.
> Буквы Т, Е, И, Я, Р, О написаны на отдельных
карточках. Ребенок берет карточки в случайном порядке и прикла-
дывает одну к друтой: а) 3 карточки; б) все 6 карточек. Какова ве-
роятность того, что получится слово: а) «ТОР»; б) «ТЕОРИЯ»?

```python
#a)
1/(sc.perm(6,3))
```

```python
#б)
1/(sc.factorial(6))
```

### Пример 1.11. 
>Используя условие примера 1.10, найти вероятность того, что получится слово «АНАНАС», если на отдельных
карточках написаны три буквы А, две буквы Н и одна буква С.

```python
n = sc.factorial(6)/(sc.factorial(3)*sc.factorial(2)*sc.factorial(1))
1/n
```

### Пример 1.16. 
> За круглым столом рассаживаются 5 мужчин и
5 женщин. Найти вероятность того, что: а) никакие два лица одного
пола не сядут рядом; б) супруги сядут рядом, если эти мужчины и
женщины образуют 5 супружеских пар.

```python
#a)
m = 2*sc.factorial(5)*sc.factorial(5)
n = sc.factorial(10)
m/n
```

```python
#b)
m = sc.factorial(5)*2**5
m/n
```

### Пример 1.17. 
>В купейный вагон (9 купе по 4 места) семи пассажирам продано 7 билетов. Найти вероятности того, что пассажиры попали: а) в два купе; б) в семь купе; в) в три купе.

```python
#b)
n = sc.comb(36,7)
m = sc.comb(9,7)*4**7
m/n
```
