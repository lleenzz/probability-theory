### 1.48.
> В старинной игре в кости необходимо было для выигрыша
получить при бросании трех игральных костей сумму очков, превосходящую 10. Найти вероятности: а) выпадения 11 очков; б) выигрыша.
* Рассмотрим событие А - выпало 11 очков
* Рассмотрим событие B - выпало 11 и более очков очков

```python
#a)
from itertools import product

ls = list(product("123456", repeat=3))
c = 0
for i in range(len(ls)):
    if sum(map(int, ls[i])) == 11:
        c += 1
c/6**3
```

```python
#b)
c = 0
for i in range(len(ls)):
    if sum(map(int, ls[i])) >= 11:
        c += 1
c/6**3
```
